```python
import tkinter as tk
from tkinter import ttk, messagebox, font as tkFont
import threading
import re
import urllib.request
import io
import os
import sys

# ─── Logo SVG embedded as base64 (we'll draw it via canvas fallback) ──────────
LOGO_B64 = ""  # Surgeon Medical Services logo

# ─── Color Palette (matches logo: navy, gold, white, crimson) ────────────────
C_NAVY      = "#1B3A5C"
C_NAVY_DARK = "#0F2438"
C_GOLD      = "#C9A84C"
C_GOLD_LT   = "#E8C97A"
C_WHITE     = "#F5F7FA"
C_CRIMSON   = "#8B1A1A"
C_STEEL     = "#4A7FA5"
C_STEEL_LT  = "#6FA3C7"
C_BG        = "#0D1F33"
C_PANEL     = "#162D47"
C_CARD      = "#1E3A56"
C_TEXT      = "#E8EDF5"
C_TEXT_DIM  = "#8FA8BF"
C_SUCCESS   = "#2ECC71"
C_WARN      = "#F39C12"


# ══════════════════════════════════════════════════════════════════════════════
#  BACKEND — NLP Pipeline (mirrors the notebook exactly)
# ══════════════════════════════════════════════════════════════════════════════

class MedicalSearchEngine:
    """Encapsulates the full preprocessing + FAISS + reranker pipeline."""

    def __init__(self, status_cb=None):
        self.status_cb = status_cb or (lambda msg, color=None: None)
        self.model       = None
        self.index       = None
        self.reranker    = None
        self.data        = None
        self.ready       = False

    # ── helpers ───────────────────────────────────────────────────────────────
    def _log(self, msg, color=None):
        self.status_cb(msg, color)

    # ── data loading ──────────────────────────────────────────────────────────
    def load_from_mongo(self, host="mongodb://localhost:27017/",
                        db_name="medical_data", col_name="nhs_diseases"):
        from pymongo import MongoClient
        import pandas as pd
        self._log("Connecting to MongoDB …")
        client = MongoClient(host)
        db = client[db_name]
        col = db[col_name]
        self.data = pd.DataFrame(list(col.find()))
        self._log(f"Loaded {len(self.data)} records from MongoDB.", C_SUCCESS)
        return self

    # ── preprocessing ─────────────────────────────────────────────────────────
    def preprocess(self):
        import nltk
        import pandas as pd
        from nltk.tokenize import word_tokenize
        from nltk.corpus import stopwords
        from nltk.stem import WordNetLemmatizer

        self._log("Downloading NLTK data …")
        for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet"]:
            nltk.download(pkg, quiet=True)

        stop_words  = set(stopwords.words("english"))
        lemmatizer  = WordNetLemmatizer()

        def flatten_sections(sections):
            text = []
            for s in (sections or []):
                title = s.get("title", "")
                paras = " ".join(s.get("paragraphs", []))
                text.append(title + " " + paras)
            return " ".join(text)

        def clean_text(text):
            text = text.lower()
            text = re.sub(r'\s+', ' ', text)
            return text.strip()

        def preprocess_text(text):
            tokens = word_tokenize(text)
            tokens = [lemmatizer.lemmatize(t)
                      for t in tokens
                      if t not in stop_words and len(t) > 2]
            return " ".join(tokens)

        self._log("Flattening sections …")
        self.data["full_text"]      = self.data["sections"].apply(flatten_sections)
        self._log("Cleaning text …")
        self.data["clean_text"]     = self.data["full_text"].apply(clean_text)
        self._log("Lemmatizing & removing stopwords …")
        self.data["processed_text"] = self.data["clean_text"].apply(preprocess_text)
        self._log("Preprocessing complete.", C_SUCCESS)
        return self

    # ── embedding ─────────────────────────────────────────────────────────────
    def build_embeddings(self):
        from sentence_transformers import SentenceTransformer
        import numpy as np
        self._log("Loading SentenceTransformer (all-MiniLM-L6-v2) …")
        self.model = SentenceTransformer("all-MiniLM-L6-v2")
        self._log("Encoding documents …")
        self.embeddings = self.model.encode(
            self.data["processed_text"].tolist(),
            show_progress_bar=False
        )
        self.embeddings = np.array(self.embeddings).astype("float32")
        self._log(f"Embeddings built: {self.embeddings.shape}", C_SUCCESS)
        return self

    # ── FAISS index ───────────────────────────────────────────────────────────
    def build_index(self):
        import faiss
        dim = self.embeddings.shape[1]
        self.index = faiss.IndexFlatL2(dim)
        self.index.add(self.embeddings)
        self._log(f"FAISS index built — {self.index.ntotal} vectors.", C_SUCCESS)
        return self

    # ── reranker ──────────────────────────────────────────────────────────────
    def build_reranker(self):
        import numpy as np
        import tensorflow as tf

        self._log("Building training pairs for reranker …")
        pairs, labels = [], []
        for _, row in self.data.iterrows():
            dt = row["processed_text"]
            pairs.append((dt, dt)); labels.append(1)
            neg = self.data.sample(1)["processed_text"].values[0]
            pairs.append((dt, neg));  labels.append(0)

        def pair_emb(q, d):
            qv = self.model.encode(q)
            dv = self.model.encode(d)
            return np.concatenate([qv, dv])

        self._log("Computing pair embeddings (this may take a while) …")
        X = np.array([pair_emb(q, d) for q, d in pairs])
        y = np.array(labels)

        self._log("Training neural reranker …")
        self.reranker = tf.keras.Sequential([
            tf.keras.layers.Dense(256, activation="relu", input_shape=(768,)),
            tf.keras.layers.Dropout(0.3),
            tf.keras.layers.Dense(64, activation="relu"),
            tf.keras.layers.Dense(1, activation="sigmoid"),
        ])
        self.reranker.compile(optimizer="adam",
                              loss="binary_crossentropy",
                              metrics=["accuracy"])
        self.reranker.fit(X, y, epochs=10, batch_size=16, verbose=0)
        self._log("Reranker trained.", C_SUCCESS)
        self._pair_emb = pair_emb
        return self

    # ── search methods ────────────────────────────────────────────────────────
    def search_faiss(self, query, top_k=5):
        import numpy as np
        qv = self.model.encode([query]).astype("float32")
        D, I = self.index.search(qv, top_k)
        return self.data.iloc[I[0]][["name"]].copy()

    def search_hybrid(self, query, k=5):
        import numpy as np
        qv = self.model.encode([query]).astype("float32")
        D, I = self.index.search(qv, 10)
        candidates = self.data.iloc[I[0]].copy()
        scores = []
        for _, row in candidates.iterrows():
            vec = self._pair_emb(query, row["processed_text"])
            score = float(self.reranker.predict(vec.reshape(1, -1), verbose=0)[0][0])
            scores.append(score)
        candidates["rerank_score"] = scores
        candidates = candidates.sort_values("rerank_score", ascending=False)
        return candidates[["name", "rerank_score"]].head(k)

    # ── full init pipeline ────────────────────────────────────────────────────
    def initialize(self):
        try:
            self.load_from_mongo()
            self.preprocess()
            self.build_embeddings()
            self.build_index()
            self.build_reranker()
            self.ready = True
            self._log("✦ System ready — enter symptoms to search.", C_GOLD_LT)
        except Exception as e:
            self._log(f"ERROR: {e}", C_CRIMSON)
            raise


# ══════════════════════════════════════════════════════════════════════════════
#  GUI — Surgeon Medical Services
# ══════════════════════════════════════════════════════════════════════════════

class SurgeonApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Surgeon Medical Services — Disease Search")
        self.geometry("980x780")
        self.minsize(820, 680)
        self.configure(bg=C_BG)

        self.engine = MedicalSearchEngine(status_cb=self._on_status)
        self._build_ui()
        self._start_engine()

    # ── status callback (thread-safe) ────────────────────────────────────────
    def _on_status(self, msg, color=None):
        color = color or C_TEXT_DIM
        self.after(0, lambda: self._append_log(msg, color))

    def _append_log(self, msg, color):
        self.log_text.config(state="normal")
        self.log_text.insert("end", f"  {msg}\n", ("colored",))
        self.log_text.tag_configure("colored", foreground=color)
        self.log_text.see("end")
        self.log_text.config(state="disabled")

    # ── UI construction ───────────────────────────────────────────────────────
    def _build_ui(self):
        # ── header ──────────────────────────────────────────────────────────
        header = tk.Frame(self, bg=C_NAVY_DARK, height=110)
        header.pack(fill="x")
        header.pack_propagate(False)

        # Logo canvas (SVG-style drawn logo)
        logo_canvas = tk.Canvas(header, width=90, height=90,
                                bg=C_NAVY_DARK, highlightthickness=0)
        logo_canvas.pack(side="left", padx=18, pady=10)
        self._draw_logo(logo_canvas)

        # Title block
        title_frame = tk.Frame(header, bg=C_NAVY_DARK)
        title_frame.pack(side="left", padx=5)

        tk.Label(title_frame, text="SURGEON", bg=C_NAVY_DARK,
                 fg=C_GOLD, font=("Georgia", 26, "bold")).pack(anchor="w")
        tk.Label(title_frame, text="Medical Services", bg=C_NAVY_DARK,
                 fg=C_STEEL_LT, font=("Georgia", 13, "italic")).pack(anchor="w")
        tk.Label(title_frame, text="طبيب جراح  •  Disease Intelligence System",
                 bg=C_NAVY_DARK, fg=C_TEXT_DIM,
                 font=("Georgia", 9)).pack(anchor="w")

        # Status indicator pill
        self.status_pill = tk.Label(header, text="⬤  Initializing …",
                                    bg=C_NAVY, fg=C_WARN,
                                    font=("Courier", 10, "bold"),
                                    padx=14, pady=6, relief="flat",
                                    bd=0, cursor="arrow")
        self.status_pill.pack(side="right", padx=22, pady=35)

        # Gold divider
        tk.Frame(self, bg=C_GOLD, height=2).pack(fill="x")

        # ── main content area ────────────────────────────────────────────────
        content = tk.Frame(self, bg=C_BG)
        content.pack(fill="both", expand=True, padx=20, pady=16)

        # Left column — search + results
        left = tk.Frame(content, bg=C_BG)
        left.pack(side="left", fill="both", expand=True, padx=(0, 10))

        # Search mode selector
        mode_frame = tk.Frame(left, bg=C_PANEL, bd=0)
        mode_frame.pack(fill="x", pady=(0, 10))

        tk.Label(mode_frame, text="Search Mode:", bg=C_PANEL,
                 fg=C_TEXT_DIM, font=("Courier", 9)).pack(side="left", padx=12, pady=8)

        self.search_mode = tk.StringVar(value="hybrid")
        for label, val in [("FAISS Vector", "faiss"), ("Hybrid + Reranker", "hybrid")]:
            rb = tk.Radiobutton(mode_frame, text=label, variable=self.search_mode,
                                value=val, bg=C_PANEL, fg=C_GOLD,
                                selectcolor=C_NAVY, activebackground=C_PANEL,
                                activeforeground=C_GOLD_LT,
                                font=("Courier", 9, "bold"))
            rb.pack(side="left", padx=8)

        # Query input
        query_outer = tk.Frame(left, bg=C_GOLD, bd=0)
        query_outer.pack(fill="x", pady=(0, 4))
        query_inner = tk.Frame(query_outer, bg=C_CARD, bd=0)
        query_inner.pack(fill="x", padx=2, pady=2)

        self.query_var = tk.StringVar()
        self.query_entry = tk.Entry(query_inner, textvariable=self.query_var,
                                   bg=C_CARD, fg=C_WHITE, insertbackground=C_GOLD,
                                   font=("Georgia", 13), relief="flat",
                                   bd=0)
        self.query_entry.pack(side="left", fill="x", expand=True, padx=14, pady=10)
        self.query_entry.bind("<Return>", lambda e: self._search())

        placeholder = "e.g. chest pain, shortness of breath …"
        self.query_entry.insert(0, placeholder)
        self.query_entry.config(fg=C_TEXT_DIM)
        self.query_entry.bind("<FocusIn>",  lambda e: self._clear_placeholder(placeholder))
        self.query_entry.bind("<FocusOut>", lambda e: self._restore_placeholder(placeholder))

        self.search_btn = tk.Button(query_inner, text="🔍  SEARCH",
                                    command=self._search,
                                    bg=C_GOLD, fg=C_NAVY_DARK,
                                    font=("Courier", 10, "bold"),
                                    relief="flat", padx=16, pady=6,
                                    cursor="hand2",
                                    activebackground=C_GOLD_LT,
                                    activeforeground=C_NAVY_DARK)
        self.search_btn.pack(side="right", padx=6, pady=5)

        # Top-K slider
        slider_row = tk.Frame(left, bg=C_BG)
        slider_row.pack(fill="x", pady=4)
        tk.Label(slider_row, text="Results:", bg=C_BG,
                 fg=C_TEXT_DIM, font=("Courier", 9)).pack(side="left")
        self.topk_var = tk.IntVar(value=5)
        tk.Scale(slider_row, from_=1, to=10, orient="horizontal",
                 variable=self.topk_var, bg=C_BG, fg=C_GOLD,
                 troughcolor=C_PANEL, highlightthickness=0,
                 sliderrelief="flat", length=180,
                 font=("Courier", 8)).pack(side="left", padx=8)

        # Results frame with scrollbar
        tk.Label(left, text="DIAGNOSTIC RESULTS", bg=C_BG,
                 fg=C_GOLD, font=("Courier", 9, "bold"),
                 anchor="w").pack(fill="x", pady=(10, 4))

        results_outer = tk.Frame(left, bg=C_GOLD)
        results_outer.pack(fill="both", expand=True)
        results_inner = tk.Frame(results_outer, bg=C_PANEL)
        results_inner.pack(fill="both", expand=True, padx=2, pady=2)

        self.results_canvas = tk.Canvas(results_inner, bg=C_PANEL,
                                        highlightthickness=0)
        scroll_y = tk.Scrollbar(results_inner, orient="vertical",
                                command=self.results_canvas.yview)
        self.results_frame = tk.Frame(self.results_canvas, bg=C_PANEL)

        self.results_frame.bind("<Configure>",
            lambda e: self.results_canvas.configure(
                scrollregion=self.results_canvas.bbox("all")))

        self.results_canvas.create_window((0, 0), window=self.results_frame,
                                          anchor="nw")
        self.results_canvas.configure(yscrollcommand=scroll_y.set)
        self.results_canvas.pack(side="left", fill="both", expand=True)
        scroll_y.pack(side="right", fill="y")

        self._show_placeholder_results()

        # Right column — log + info
        right = tk.Frame(content, bg=C_BG, width=280)
        right.pack(side="right", fill="y")
        right.pack_propagate(False)

        tk.Label(right, text="SYSTEM LOG", bg=C_BG,
                 fg=C_GOLD, font=("Courier", 9, "bold"),
                 anchor="w").pack(fill="x", pady=(0, 4))

        log_outer = tk.Frame(right, bg=C_GOLD)
        log_outer.pack(fill="both", expand=True)
        log_inner = tk.Frame(log_outer, bg=C_NAVY_DARK)
        log_inner.pack(fill="both", expand=True, padx=2, pady=2)

        self.log_text = tk.Text(log_inner, bg=C_NAVY_DARK, fg=C_TEXT_DIM,
                                font=("Courier", 8), wrap="word",
                                relief="flat", state="disabled",
                                padx=8, pady=6)
        log_scroll = tk.Scrollbar(log_inner, command=self.log_text.yview)
        self.log_text.configure(yscrollcommand=log_scroll.set)
        self.log_text.pack(side="left", fill="both", expand=True)
        log_scroll.pack(side="right", fill="y")

        # Pipeline info card
        info = tk.Frame(right, bg=C_CARD, bd=0)
        info.pack(fill="x", pady=(10, 0))
        tk.Label(info, text="PIPELINE", bg=C_CARD,
                 fg=C_GOLD, font=("Courier", 8, "bold")).pack(anchor="w", padx=10, pady=(8,2))
        steps = [
            ("①", "MongoDB", "medical_data.nhs_diseases"),
            ("②", "NLP Prep", "clean → tokenize → lemmatize"),
            ("③", "Embeddings", "all-MiniLM-L6-v2"),
            ("④", "FAISS", "IndexFlatL2 (348 vectors)"),
            ("⑤", "Reranker", "TF Dense 256→64→1 sigmoid"),
        ]
        for icon, title, detail in steps:
            row = tk.Frame(info, bg=C_CARD)
            row.pack(fill="x", padx=10, pady=1)
            tk.Label(row, text=icon, bg=C_CARD, fg=C_GOLD,
                     font=("Courier", 8, "bold"), width=2).pack(side="left")
            tk.Label(row, text=f"{title}: ", bg=C_CARD, fg=C_STEEL_LT,
                     font=("Courier", 7, "bold")).pack(side="left")
            tk.Label(row, text=detail, bg=C_CARD, fg=C_TEXT_DIM,
                     font=("Courier", 7), wraplength=190,
                     justify="left").pack(side="left")
        tk.Frame(info, bg=C_CARD, height=8).pack()

        # Footer
        tk.Frame(self, bg=C_GOLD, height=1).pack(fill="x")
        footer = tk.Frame(self, bg=C_NAVY_DARK, height=28)
        footer.pack(fill="x")
        tk.Label(footer, text="© Surgeon Medical Services  •  طبيب جراح  •  NHS Disease Intelligence",
                 bg=C_NAVY_DARK, fg=C_TEXT_DIM,
                 font=("Courier", 8)).pack(pady=6)

    # ── Logo drawn on canvas (real PNG logo) ────────────────────────────────
    def _draw_logo(self, c):
        import base64, io
        try:
            from PIL import Image, ImageTk
            img_data = base64.b64decode(LOGO_B64)
            img = Image.open(io.BytesIO(img_data)).resize((86, 86), Image.LANCZOS)
            self._logo_img = ImageTk.PhotoImage(img)
            c.create_image(45, 45, image=self._logo_img)
        except Exception:
            # Fallback: draw a simple caduceus
            c.create_oval(4, 4, 86, 86, outline=C_GOLD, width=3, fill=C_NAVY)
            c.create_oval(12, 12, 78, 78, outline=C_WHITE, width=1, fill=C_NAVY_DARK)
            c.create_line(45, 20, 45, 72, fill=C_GOLD, width=3)
            c.create_arc(20, 18, 45, 38, start=0, extent=180, style="arc",
                         outline=C_STEEL_LT, width=2)
            c.create_arc(45, 18, 70, 38, start=0, extent=180, style="arc",
                         outline=C_STEEL_LT, width=2)
            for y in [35, 48, 61]:
                c.create_oval(36, y-5, 54, y+5, outline=C_GOLD_LT, width=1.5, fill="")
            c.create_rectangle(42, 15, 48, 27, fill=C_CRIMSON, outline="")
            c.create_rectangle(38, 18, 52, 24, fill=C_CRIMSON, outline="")

    # ── Placeholder results ──────────────────────────────────────────────────
    def _show_placeholder_results(self):
        for w in self.results_frame.winfo_children():
            w.destroy()
        tk.Label(self.results_frame,
                 text="\n\n\n  Enter symptoms above and press Search\n"
                      "  once the system finishes initializing.\n\n\n",
                 bg=C_PANEL, fg=C_TEXT_DIM,
                 font=("Georgia", 10, "italic"),
                 justify="left").pack(padx=20, pady=20)

    # ── Result cards ─────────────────────────────────────────────────────────
    def _render_results(self, df, mode):
        for w in self.results_frame.winfo_children():
            w.destroy()

        has_score = "rerank_score" in df.columns

        for rank, (_, row) in enumerate(df.iterrows(), 1):
            card = tk.Frame(self.results_frame, bg=C_CARD, bd=0)
            card.pack(fill="x", padx=8, pady=5)

            # rank badge
            badge = tk.Label(card, text=f" #{rank} ",
                             bg=C_GOLD if rank == 1 else C_NAVY,
                             fg=C_NAVY_DARK if rank == 1 else C_GOLD,
                             font=("Courier", 10, "bold"))
            badge.pack(side="left", padx=(10, 6), pady=10)

            # name
            name_lbl = tk.Label(card, text=row["name"],
                                bg=C_CARD, fg=C_WHITE if rank > 1 else C_GOLD_LT,
                                font=("Georgia", 11, "bold" if rank == 1 else "normal"),
                                anchor="w", justify="left", wraplength=380)
            name_lbl.pack(side="left", fill="x", expand=True, pady=10)

            # score bar (hybrid only)
            if has_score:
                score = float(row["rerank_score"])
                score_frame = tk.Frame(card, bg=C_CARD)
                score_frame.pack(side="right", padx=12, pady=10)
                bar_bg = tk.Frame(score_frame, bg=C_NAVY, width=80, height=6)
                bar_bg.pack()
                bar_bg.pack_propagate(False)
                bar_fill = tk.Frame(bar_bg, bg=C_STEEL_LT,
                                    width=int(80 * min(score * 2, 1.0)), height=6)
                bar_fill.place(x=0, y=0)
                tk.Label(score_frame, text=f"{score:.3f}",
                         bg=C_CARD, fg=C_STEEL_LT,
                         font=("Courier", 8)).pack()

            # gold bottom border for rank 1
            if rank == 1:
                tk.Frame(card, bg=C_GOLD, height=2).pack(fill="x")

        if len(df) == 0:
            tk.Label(self.results_frame,
                     text="No results found. Try different symptoms.",
                     bg=C_PANEL, fg=C_WARN,
                     font=("Georgia", 10, "italic")).pack(padx=20, pady=20)

    # ── Placeholder clearing ──────────────────────────────────────────────────
    def _clear_placeholder(self, placeholder):
        if self.query_entry.get() == placeholder:
            self.query_entry.delete(0, "end")
            self.query_entry.config(fg=C_WHITE)

    def _restore_placeholder(self, placeholder):
        if not self.query_entry.get():
            self.query_entry.insert(0, placeholder)
            self.query_entry.config(fg=C_TEXT_DIM)

    # ── Search ────────────────────────────────────────────────────────────────
    def _search(self):
        if not self.engine.ready:
            messagebox.showwarning("Not Ready",
                "The system is still initializing.\nPlease wait …")
            return

        placeholder = "e.g. chest pain, shortness of breath …"
        query = self.query_var.get().strip()
        if not query or query == placeholder:
            messagebox.showinfo("Input needed", "Please enter symptoms to search.")
            return

        self.search_btn.config(state="disabled", text="Searching …")
        mode = self.search_mode.get()
        k    = self.topk_var.get()

        def run():
            try:
                self._on_status(f"Searching: '{query}' [{mode}] …", C_STEEL_LT)
                if mode == "faiss":
                    df = self.engine.search_faiss(query, top_k=k)
                else:
                    df = self.engine.search_hybrid(query, k=k)
                self.after(0, lambda: self._render_results(df, mode))
                self._on_status(f"Done — {len(df)} results.", C_SUCCESS)
            except Exception as e:
                self._on_status(f"Search error: {e}", C_CRIMSON)
            finally:
                self.after(0, lambda: self.search_btn.config(
                    state="normal", text="🔍  SEARCH"))

        threading.Thread(target=run, daemon=True).start()

    # ── Engine init ───────────────────────────────────────────────────────────
    def _start_engine(self):
        def run():
            try:
                self.engine.initialize()
                self.after(0, lambda: self.status_pill.config(
                    text="⬤  System Ready", fg=C_SUCCESS))
            except Exception as e:
                self.after(0, lambda: self.status_pill.config(
                    text="⬤  Init Failed", fg=C_CRIMSON))

        threading.Thread(target=run, daemon=True).start()


# ══════════════════════════════════════════════════════════════════════════════
if __name__ == "__main__":
    app = SurgeonApp()
    app.mainloop()
```
